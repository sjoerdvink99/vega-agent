# 3 · Did the new model break the harness?

This is one spreadsheet copilot run with two models. Colour shows which model takes each path more often. The data is synthetic and produced by `data/generate.py`.

```python
import vega_agent as va

old = va.Corpus(va.read("../data/copilot-model-a.jsonl"), "model-a")
new = va.Corpus(va.read("../data/copilot-model-b.jsonl"), "model-b")
```

```python
harness = "model -> tool.read_range[] -> (model -> tool[])* -> model"
upgrade = va.Diff(
    old,
    new,
    spec=harness,
    labels=("model-a", "model-b"),
    queries={
        "retry after failed write": "tool.write_range! -> model -> tool.write_range",
        "write without read-back": "tool.write_range -> model -> tool.write_range",
        "delegates to a subagent": "agent",
    },
)
upgrade.chart(level="names")
```

The same comparison as other marks: the most common variants, and how often each step follows another.

```python
upgrade.chart(mark="variants", level="names")
```

```python
upgrade.chart(mark="transitions")
```
