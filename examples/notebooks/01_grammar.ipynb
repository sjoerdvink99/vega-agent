# 1 · The grammar produces the picture

An expression is data, and its diagram is a pure function of that data. Every expression renders itself.

```python
import vega_agent as va

va.parse("(model -> tool[])* -> model")
```

```python
patterns = {
    "routing": "model.route -> (agent.billing | agent.tech | model.answer)",
    "voting": "model.answer[5] -> code.vote",
    "orchestrator": "model.plan(=> task:text[]) -> agent.worker{ (model -> tool[])* -> model }[] -> model.synthesize",
    "evaluator": "model.draft -> (model.critique -> model.revise){1,3}",
    "approval": "(model -> human.approve? -> tool[])* -> model",
}
va.diagram(patterns["orchestrator"], title="Orchestrator–workers")
```

A chart is a declarative view spec. The same JSON renders in any web app through `vega-agent`, through the CLI, or here.

```python
va.diagram(patterns["routing"]).to_dict()
```

## Generative

`sample` draws runs from a spec, and each run conforms to it.

```python
runs = va.sample(patterns["orchestrator"], 3, seed=4)
[va.conforms(run, patterns["orchestrator"]) for run in runs]
```

## Abstraction

There are four levels, from exact to coarse.

```python
run = va.trace("model -> tool.read(q:text => r:table) -> model -> (tool.read || tool.grep!) -> model")
for level in va.LEVELS:
    print(f"{level:7}", va.abstract(run.process, level))
```
