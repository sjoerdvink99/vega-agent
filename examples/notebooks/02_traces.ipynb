## Standards in, standards out

Transcripts from OpenAI, Anthropic and Gemini read the same way as spans. Exports go to OpenTelemetry, XES, OCEL 2.0, BPMN 2.0 and process trees.

```python
va.read(
    [
        {"role": "user", "content": "Weather in Delft?"},
        {
            "role": "assistant",
            "content": [{"type": "tool_use", "id": "t1", "name": "get_weather", "input": {}}],
        },
        {
            "role": "user",
            "content": [{"type": "tool_result", "tool_use_id": "t1", "content": "rain"}],
        },
        {"role": "assistant", "content": "Rain."},
    ]
)[0]
```

```python
print(va.to_process_tree("(model -> tool[])* -> model"))
print(va.to_bpmn("model -> (tool.a || tool.b) -> model")[:400])
```

```python
print(va.to_xes(runs[:1])[:600])
```
